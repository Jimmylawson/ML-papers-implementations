```python
ds = load_dataset("Helsinki-NLP/opus_books", "en-it", split="train")
print("Dataset size:", len(ds))
print(ds[0])
```

```python
example = ds[0]
english = example["translation"]["en"]
italian = example["translation"]["it"]
print("English",english)
print("Italian",italian)
```

```python
from train import get_or_build_tokenizer
from config import get_config
config = get_config()
tokenizer_en = get_or_build_tokenizer(config,ds,"en")
sentence = "I love machine learning"
encoded = tokenizer_en.encode(sentence)

print("Token IDs:", encoded.ids)
print("Tokens", encoded.tokens)
print("Vocabulary size", tokenizer_en.get_vocab_size())
```

```python
sentence_it = "Amo imparare cose nuove"
tokenizer_it = get_or_build_tokenizer(config,ds,"it")
encoded_it = tokenizer_it.encode(sentence_it)

print("Italian tokens:", encoded_it.tokens)
print("Italian token IDs:", encoded_it.ids)

decoded_it = tokenizer_it.decode(encoded_it.ids)
print("Decoded Italian:", decoded_it)
```

```python
from dataset import BilingualDataset

train_ds = BilingualDataset(
    ds,
    tokenizer_en,
    tokenizer_it,
    "en",
    "it",
    128
)

sample = train_ds[0]

print("English:", sample["src_text"])
print("Italian:", sample["tgt_text"])

print("Encoder:", sample["encoder_input"].shape)
print("Decoder:", sample["decoder_input"].shape)
print("Label:", sample["label"].shape)
print("Encoder mask:", sample["encoder_mask"].shape)
print("Decoder mask:", sample["decoder_mask"].shape)
```
